```python
from reusing_matrices import Matrix
import math
```

```python

# Exercise 1

x = Matrix([[math.log(2), 1],[math.log(4),1/3],[math.log(5),0.25]])
y = Matrix([[0],[-1],[5]])

x_t = Matrix([[math.log(2), 1],[math.log(4),1/3],[math.log(5),0.25]])
x_t.transpose()

xtx = x_t.matrix_multiply(x)
xtx.inverse_matrix_via_rref()

x_t_for_right = Matrix([[math.log(2), 1],[math.log(4),1/3],[math.log(5),0.25]])
x_t_for_right.transpose()

pseudoinverse = xtx.matrix_multiply(x_t_for_right)
#pseudoinverse.show()

pseudoinverse.matrix_multiply(y)
pseudoinverse.show()
```
