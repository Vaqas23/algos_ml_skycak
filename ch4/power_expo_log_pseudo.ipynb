```python
from reusing_matrices import Matrix
import math

# Exercise 1 [(1, 0.2),(2, 0.3),(3, 0.5)]

x = Matrix([[1,math.log(1)],[1,math.log(2)],[1,math.log(3)]])
y = Matrix([[math.log(0.2)],[math.log(0.3)],[math.log(0.5)]])

x_t = Matrix([[1,math.log(1)],[1,math.log(2)],[1,math.log(3)]])
x_t.transpose()

xtx = x_t.matrix_multiply(x)
xtx.inverse_matrix_via_rref()

x_t_for_right = Matrix([[1,math.log(1)],[1,math.log(2)],[1,math.log(3)]])
x_t_for_right.transpose()

pseudoinverse = xtx.matrix_multiply(x_t_for_right)
#pseudoinverse.show()

pseudoinverse.matrix_multiply(y)
pseudoinverse.show()

```
