```python
# Data

data = [(0, 0),(1, 10),(2, 20),(3, 50),(4, 35),(5, 100),
(6, 110),(7, 190),(8, 150),(9, 260),(10, 270)]

from reusing_matrices import Matrix
```

```python
# Linear Model

rss = 0 # cross-validated residual sum of squares


for i in range(len(data)):

    x_data = []
    y_data = []

    for j in range(len(data)):

        if j != i:
            x_data.append([data[j][0],1]) #append coefficient of m (x value) and the coefficient of b (1)
            y_data.append([data[j][1]]) #append y value
    

    x = Matrix(x_data)
    y = Matrix(y_data)

    x_t = Matrix(x_data)
    x_t.transpose()

    xtx = x_t.matrix_multiply(x)
    xtx.inverse_matrix_via_rref()

    x_t_for_right = Matrix(x_data)
    x_t_for_right.transpose()

    pseudoinverse = xtx.matrix_multiply(x_t_for_right)

    pseudoinverse.matrix_multiply(y)

    # psuedo inverse now have our m,b values
    m = pseudoinverse.matrix[0][0]
    b = pseudoinverse.matrix[1][0]
    x_val = data[i][0]

    predicted_val = m * x_val + b
    real_val = data[i][1]
    rss += (real_val - predicted_val)**2


print(f"RSS: {rss}")
        
```

```python
# Quadratic Model

rss = 0 # cross-validated residual sum of squares

for i in range(len(data)):

    x_data = []
    y_data = []

    for j in range(len(data)):

        if j != i:
            x_data.append([(data[j][0])**2,data[j][0],1]) #append coefficient of m (x value) and the coefficient of b (1)
            y_data.append([data[j][1]]) #append y value
    

    x = Matrix(x_data)
    y = Matrix(y_data)

    x_t = Matrix(x_data)
    x_t.transpose()

    xtx = x_t.matrix_multiply(x)
    xtx.inverse_matrix_via_rref()

    x_t_for_right = Matrix(x_data)
    x_t_for_right.transpose()

    pseudoinverse = xtx.matrix_multiply(x_t_for_right)
    pseudoinverse.matrix_multiply(y)

    # psuedo inverse now have our a,b,c values
    a = pseudoinverse.matrix[0][0]
    b = pseudoinverse.matrix[1][0]
    c = pseudoinverse.matrix[2][0]
    x_val = data[i][0]

    predicted_val = a * x_val**2 + b * x_val + c
    real_val = data[i][1]
    rss += (real_val - predicted_val)**2


print(f"RSS: {rss}")
        
```

```python
# 8th Degree Model

rss = 0 # cross-validated residual sum of squares

for i in range(len(data)):

    x_data = []
    y_data = []

    for j in range(len(data)):

        if j != i:
            x_data.append([(data[j][0])**8,(data[j][0])**7,(data[j][0])**6,(data[j][0])**5,(data[j][0])**4,(data[j][0])**3,(data[j][0])**2,(data[j][0]),1]) #append coefficient of m (x value) and the coefficient of b (1)
            y_data.append([data[j][1]]) #append y value
    

    x = Matrix(x_data)
    y = Matrix(y_data)

    x_t = Matrix(x_data)
    x_t.transpose()

    xtx = x_t.matrix_multiply(x)
    xtx.inverse_matrix_via_rref()

    x_t_for_right = Matrix(x_data)
    x_t_for_right.transpose()

    pseudoinverse = xtx.matrix_multiply(x_t_for_right)
    pseudoinverse.matrix_multiply(y)

    # psuedo inverse now have our a-h,k values. Im aware this makes less efficient, but readability is key for me.
    a = pseudoinverse.matrix[0][0]
    b = pseudoinverse.matrix[1][0]
    c = pseudoinverse.matrix[2][0]
    d = pseudoinverse.matrix[3][0]
    e = pseudoinverse.matrix[4][0]
    f = pseudoinverse.matrix[5][0]
    g = pseudoinverse.matrix[6][0]
    h = pseudoinverse.matrix[7][0]
    k = pseudoinverse.matrix[8][0] #skip i/j, not gonna reuse from loop
    x_val = data[i][0]

    predicted_val = (a * x_val**8) + (b * x_val**7) + (c * x_val**6) + (d * x_val**5) + (e * x_val**4) + (f * x_val**3) + (g * x_val**2) + (h * x_val) + k
    real_val = data[i][1]
    rss += (real_val - predicted_val)**2


print(f"RSS: {rss}")
        
```
