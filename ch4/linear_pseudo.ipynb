```python
""" 
The pseudoinverse of a matrix is defined as: 

Inverse(Transpose of Matrix X * Matrix X) * Transpose of Matrix X

This form applies to tall matrices with full column rank.

"""

import matplotlib.pyplot as plt
from reusing_matrices import Matrix
```

```python
"""
Example 1

[(1, 0),(3, -1),(4, 5)]

Inverse(Transpose of Matrix X * Matrix X) * Transpose of Matrix X * y matrix  = the answer

Matrix X is made up of the coefficients of m and b, after plugging in the x values.

This is a linear fit

"""

x = Matrix([[1, 1], [3, 1], [4, 1]])
y = Matrix([[0],[-1],[5]])

x_t = Matrix([[1, 0], [3, -1], [4, 5]])
x_t.transpose()

xtx = x_t.matrix_multiply(x)
xtx.inverse_matrix_via_rref()

x_t_for_right = Matrix([[1, 0], [3, -1], [4, 5]])
x_t_for_right.transpose()

pseudoinverse = xtx.matrix_multiply(x_t_for_right)
#pseudoinverse.show()

pseudoinverse.matrix_multiply(y)
pseudoinverse.show()


# y = 4.375x -12.09375 is the answers

```

```python
#  [(−2, 3),(1, 0),(3, −1),(4, 5)]
# linear fit

x = Matrix([[-2, 1], [1, 1], [3, 1], [4,1]])
y = Matrix([[3],[0],[-1],[5]])

x_t = Matrix([[-2, 1], [1, 1], [3, 1], [4,1]])
x_t.transpose()

xtx = x_t.matrix_multiply(x)
xtx.inverse_matrix_via_rref()

x_t_for_right = Matrix([[-2, 1], [1, 1], [3, 1], [4,1]])
x_t_for_right.transpose()

pseudoinverse = xtx.matrix_multiply(x_t_for_right)
#pseudoinverse.show()

pseudoinverse.matrix_multiply(y)
pseudoinverse.show()

# y = 0.023809523809523836x + 1.7142857142857142
```
